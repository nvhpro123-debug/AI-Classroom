## a) Một vòng lặp K-Means

### Tính khoảng cách

| Điểm | d(P,C1) | d(P,C2) | Cụm |
|---|---|---|---|
| P1 | √5 ≈ 2.236 | √53 ≈ 7.280 | C1 |
| P2 | √5 ≈ 2.236 | √29 ≈ 5.385 | C1 |
| P3 | 3 | 3 | C1 |
| P4 | √50 ≈ 7.071 | √2 ≈ 1.414 | C2 |
| P5 | √26 ≈ 5.099 | √2 ≈ 1.414 | C2 |
| P6 | 8 | 2 | C2 |

### Giải thích điểm P3

P3 có:

```text
d(P3,C1) = d(P3,C2) = 3
```

Theo `np.argmin`, khi có nhiều giá trị nhỏ nhất bằng nhau thì sẽ chọn chỉ số nhỏ hơn trước.

Do đó:

```text
P3 thuộc cụm C1
```

# PHẦN B – K-NN

## a) Tính khoảng cách đến q = (0,0)

Ta dùng bình phương khoảng cách để dễ tính.

| Điểm | Tọa độ | Nhãn | d² |
|---|---|---|---|
| T1 | (1,1) | Lớp 1 | 2 |
| T2 | (2,1) | Lớp 0 | 5 |
| T3 | (2,2) | Lớp 1 | 8 |
| T4 | (1,3) | Lớp 0 | 10 |
| T5 | (2,3) | Lớp 0 | 13 |
| T6 | (3,3) | Lớp 1 | 18 |

### Thứ tự tăng dần

```text
T1 → T2 → T3 → T4 → T5 → T6
```

---

## b) Với k = 3

3 điểm gần nhất:

```text
T1(Lớp1), T2(Lớp0), T3(Lớp1)
```

Số phiếu:

| Lớp | Phiếu |
|---|---|
| Lớp 1 | 2 |
| Lớp 0 | 1 |

→ q được phân vào:

```text
Lớp 1
```

---

## c) Với k = 4

4 điểm gần nhất:

```text
T1(Lớp1), T2(Lớp0), T3(Lớp1), T4(Lớp0)
```

Kết quả:

| Lớp | Phiếu |
|---|---|
| Lớp 1 | 2 |
| Lớp 0 | 2 |

Hòa phiếu.

Trong mã:

````python
classes = {'Lớp 1': 2, 'Lớp 0': 2}
max(classes, key=classes.get)
````

Python sẽ trả về phần tử xuất hiện trước trong dictionary khi giá trị bằng nhau.

Do lớp được thêm theo thứ tự duyệt láng giềng:

```text
Lớp 1 được thêm trước
```

nên kết quả:

```text
q thuộc Lớp 1
```

---

### Với k = 5

5 điểm gần nhất:

```text
T1(L1), T2(L0), T3(L1), T4(L0), T5(L0)
```

Số phiếu:

| Lớp | Phiếu |
|---|---|
| Lớp 0 | 3 |
| Lớp 1 | 2 |

→ Kết quả đổi thành:

```text
Lớp 0
```

Vì lớp 0 có đa số phiếu.
